# Step 2: Feature Engineering — Crypto Market Regime Monitor

This notebook loads OHLCV candle data from `data/market_data.db` (populated by `src/ingest.py`)
and engineers features used to classify market regime (trending / ranging / high-volatility).

**Data source note:** run `python src/ingest.py --symbol BTCUSDT --interval 15m --limit 500` first to populate the database with live data from Binance's public API.


```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("../src")
from features import compute_features

pd.set_option('display.max_columns', None)
%matplotlib inline
```

## Load raw OHLCV data

```python
DB_PATH = "../data/market_data.db"
SYMBOL = "btcusdt"

conn = sqlite3.connect(DB_PATH)
df = pd.read_sql(f"SELECT * FROM {SYMBOL} ORDER BY open_time", conn)
conn.close()

df["datetime"] = pd.to_datetime(df["open_time"], unit="ms")
df = df.set_index("datetime").sort_index()

print(df.shape)
df.head()
```

```python
df["close"].plot(figsize=(12, 4), title=f"{SYMBOL.upper()} Close Price")
plt.ylabel("Price")
plt.show()
```

## Feature engineering

We engineer features that describe **how the market is behaving** rather than trying to predict
the next price directly (raw price prediction on crypto is close to a random walk — weak signal).

Features:
- **Returns** — log returns, the base unit for volatility/momentum calculations
- **Rolling volatility** — realized volatility over multiple windows (short vs long)
- **Moving averages & crossovers** — trend direction signal
- **RSI** — momentum / overbought-oversold indicator
- **Volume anomaly** — spikes relative to rolling average volume
- **ATR (Average True Range)** — volatility that accounts for intra-candle range, not just close-to-close


```python
# compute_features() is imported from src/features.py — the same function
# the live prediction pipeline (Step 4) will use, so there's no drift between
# training-time and inference-time feature logic.

features_df = compute_features(df)
features_df = features_df.dropna().copy()
print(features_df.shape)
features_df[["close", "vol_short", "vol_long", "trend_strength", "rsi", "volume_ratio", "atr_pct"]].tail()
```

```python
fig, axes = plt.subplots(4, 1, figsize=(12, 10), sharex=True)

axes[0].plot(features_df.index, features_df["close"])
axes[0].set_title("Close Price")

axes[1].plot(features_df.index, features_df["vol_short"], label="short vol")
axes[1].plot(features_df.index, features_df["vol_long"], label="long vol")
axes[1].legend()
axes[1].set_title("Rolling Volatility")

axes[2].plot(features_df.index, features_df["trend_strength"])
axes[2].axhline(0, color="gray", linestyle="--", linewidth=0.8)
axes[2].set_title("Trend Strength (MA crossover, normalized)")

axes[3].plot(features_df.index, features_df["rsi"])
axes[3].axhline(70, color="red", linestyle="--", linewidth=0.8)
axes[3].axhline(30, color="green", linestyle="--", linewidth=0.8)
axes[3].set_title("RSI")

plt.tight_layout()
plt.show()
```

## Save engineered features

We persist this to a new table so the modeling notebook can load it directly without
recomputing features every time.


```python
conn = sqlite3.connect(DB_PATH)
features_df.to_sql(f"{SYMBOL}_features", conn, if_exists="replace", index=True)
conn.close()
print("Saved engineered features to database.")
```
