```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/market_data.db")
df = pd.read_sql("SELECT * FROM btcusdt ORDER BY open_time", conn)
conn.close()

df["datetime"] = pd.to_datetime(df["open_time"], unit="ms")
print(df.shape)
print(df[["datetime", "open", "high", "low", "close", "volume"]].tail())
```
